```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_1=pd.read_csv("C:\\Users\\rohit\\Desktop\\internship datasets\\Task_2\\Unemployment in India.csv")
data_1
```

```python
data_1.head()
```

```python
data_1.shape
```

```python
data_1.columns
```

```python
data_1.columns=data_1.columns.str.strip()
data_1.columns
```

```python
data_1.isnull().sum()
```

```python
data_1.duplicated().sum()
```

```python
data_1=data_1.dropna()
data_1.shape
```

```python
data_1=data_1.drop_duplicates()
data_1
```

```python
data_1.isnull().sum()
data_1.duplicated().sum()
```

```python
data_1["Date"]=pd.to_datetime(data_1["Date"],dayfirst=True)
data_1.head()
```

```python
average=data_1["Estimated Unemployment Rate (%)"].mean()
print("Average Unemployment Rate:",round(average,2),"%")
```

```python
region_rate=data_1.groupby("Region")["Estimated Unemployment Rate (%)"].mean()
region_rate
```

```python
trend=data_1.groupby("Date")["Estimated Unemployment Rate (%)"].mean()
trend
```

```python
plt.figure(figsize=(12,5))
region_rate.plot(kind='bar')
plt.title('Average Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(trend.index, trend.values)
plt.title('Unemployment Rate Trend')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.show()
```

```python
data_2=pd.read_csv("C:\\Users\\rohit\\Desktop\\internship datasets\\Task_2\\Unemployment_Rate_upto_11_2020.csv")
data_2
data_2.head()
```

```python
data_2.shape
```

```python
data_2.columns=data_2.columns.str.strip()
data_2.isnull().sum()
```

```python
data_2["Date"]=pd.to_datetime(data_2["Date"],dayfirst=True)
data_2.head()
```

```python
covid_trend=data_2.groupby("Date")["Estimated Unemployment Rate (%)"].mean()
covid_trend
```

```python
plt.figure(figsize=(12,5))

plt.plot(covid_trend.index, covid_trend.values, marker='o')

plt.title('Unemployment Rate During COVID-19 Period')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')

plt.xticks(rotation=45)

plt.show()
```

```python
highest=data_2.loc[data_2['Estimated Unemployment Rate (%)'].idxmax()]
highest
```

```python
lowest=data_2.loc[data_2['Estimated Unemployment Rate (%)'].idxmin()]
lowest
```

```python
print("Highest Unemployment Rate:",highest['Estimated Unemployment Rate (%)'],"%")
print("Region:", highest['Region'])
print("Date:", highest['Date'])
```

```python
print("Lowest Unemployment Rate:",lowest['Estimated Unemployment Rate (%)'],"%")
print("Region:", lowest['Region'])
print("Date:", lowest['Date'])
```

```python
## Key Findings

# The average unemployment rate in the dataset was approximately 11.79%.
# Unemployment rates varied significantly across different regions.
# The COVID-19 period showed a significant increase in unemployment.
# May had the highest average monthly unemployment rate, while November had the lowest.
# Employment generation, skill development, and financial support can help highly affected regions.

```
